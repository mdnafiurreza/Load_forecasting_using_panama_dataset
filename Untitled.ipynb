```python
# --- Imports ---
import pandas as pd
import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader

# --- Load Dataset ---
df = pd.read_csv("updated_dataset.csv")

# --- Feature and Target Selection ---
feature_cols = [
    'holiday', 'school', 'avg_T2M', 'avg_QV2M', 'avg_TQL', 'avg_W2M',
    'season', 'hour_of_day', 'is_weekend', 'year', 'month', 'day_of_week',
    'nat_demand_1hr_lag', 'nat_demand_24hr_lag', 'nat_demand_1week_lag'
]
X = df[feature_cols].values
y = df['nat_demand'].values

# --- Create Time Windows ---
def create_windows(X, y, window_size=24):
    X_seq, y_seq = [], []
    for i in range(len(X) - window_size):
        X_seq.append(X[i:i+window_size])
        y_seq.append(y[i+window_size])
    return np.array(X_seq), np.array(y_seq)

X_seq, y_seq = create_windows(X, y, window_size=24)

# Reshape for CNN: (samples, 1, time_steps * features)
X_seq = X_seq.reshape((X_seq.shape[0], 1, -1))

# --- Convert to Torch Tensors ---
X_tensor = torch.tensor(X_seq, dtype=torch.float32)
y_tensor = torch.tensor(y_seq, dtype=torch.float32).unsqueeze(1)

# --- Train-Test Split ---
split = int(len(X_tensor) * 0.8)
X_train, X_test = X_tensor[:split], X_tensor[split:]
y_train, y_test = y_tensor[:split], y_tensor[split:]

# --- DataLoader for Training ---
train_ds = TensorDataset(X_train, y_train)
train_dl = DataLoader(train_ds, batch_size=64, shuffle=True)

# --- Define CNN Model ---
class TimeWindowCNN(nn.Module):
    def __init__(self, input_len):
        super().__init__()
        self.conv1 = nn.Conv1d(1, 32, kernel_size=3, padding=1)
        self.bn1 = nn.BatchNorm1d(32)
        self.pool = nn.MaxPool1d(2)
        self.conv2 = nn.Conv1d(32, 64, kernel_size=3, padding=1)
        self.bn2 = nn.BatchNorm1d(64)
        self.drop = nn.Dropout(0.5)
        self.fc1 = nn.Linear((input_len // 2) * 64, 128)
        self.fc2 = nn.Linear(128, 1)
        self.relu = nn.ReLU()

    def forward(self, x):
        x = self.relu(self.bn1(self.conv1(x)))
        x = self.pool(x)
        x = self.relu(self.bn2(self.conv2(x)))
        x = x.view(x.size(0), -1)
        x = self.drop(self.relu(self.fc1(x)))
        out = self.fc2(x)
        return out

# --- Instantiate Model ---
device = torch.device("cpu")
cnn_model = TimeWindowCNN(input_len=X_tensor.shape[2]).to(device)

# --- Loss and Optimizer ---
loss_fn = nn.MSELoss()
optimizer = torch.optim.Adam(cnn_model.parameters(), lr=1e-3)
scheduler = torch.optim.lr_scheduler.StepLR(optimizer, step_size=10, gamma=0.2)

# --- Training Loop ---
epochs = 20
cnn_model.train()
for epoch in range(epochs):
    total_loss = 0
    for xb, yb in train_dl:
        optimizer.zero_grad()
        preds = cnn_model(xb)
        loss = loss_fn(preds, yb)
        loss.backward()
        optimizer.step()
        total_loss += loss.item() * xb.size(0)
    scheduler.step()
    avg_loss = total_loss / len(train_dl.dataset)
    print(f"Epoch {epoch+1}/{epochs} – Loss: {avg_loss:.4f}")

# --- Save Features If Needed ---
# After training, you can use cnn_model to extract features if you want to build hybrid models!
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# --- Evaluation ---
cnn_model.eval()
with torch.no_grad():
    y_pred = cnn_model(X_test).squeeze().numpy()
    y_true = y_test.squeeze().numpy()

# --- Compute Metrics ---
mae  = mean_absolute_error(y_true, y_pred)
mse  = mean_squared_error(y_true, y_pred)
r2   = r2_score(y_true, y_pred)
mape = np.mean(np.abs((y_true - y_pred) / y_true))

# --- Print Results ---
print(f"\nCNN Test Results:")
print(f"R²    : {r2:.4f}")
print(f"MAE   : {mae:.2f}")
print(f"MSE   : {mse:.2f}")
print(f"MAPE  : {mape:.4f}")
```
